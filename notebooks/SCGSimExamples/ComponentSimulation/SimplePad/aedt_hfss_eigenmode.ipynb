# SimplePad HFSS：本徵模態與表面 EPR

以正方形焊墊（SimplePad）示範 HFSS
本徵模態（Eigenmode）與表面能量參與率（Surface
EPR）。依序使用三份獨立筆記本，每一步都能在全新核心（Kernel）開始。

| 步驟 | 筆記本 | 輸入與輸出 |
|------------------------|------------------------|------------------------|
| 1 | [準備交接包（Prepare Handoff）](aedt_hfss_eigenmode_prepare.qmd) | 版圖、堆疊、物理與數值設定 → 新交接包 |
| 2 | [執行（Run）](aedt_hfss_eigenmode_run.qmd) | 已準備的交接包 → 原生求解與回傳檔案 |
| 3 | [分析（Analyze）](aedt_hfss_eigenmode_analyze.qmd) | 明確指定的回傳資料夾 → 模態、原生回合、EPR 與計算成本 |

第一步保留 200 µm 焊墊、20 µm 間隙、1000 µm 晶片、10 nH
線性接面，以及原本的模態搜尋與自適應設定。執行與分析不重新建立這些輸入。

安裝指令及公開版本見 Prepare Handoff。已保存場的授權 AEDT
評估與離線薄膜損耗重分析，分別列在 Analyze
的兩個選用章節；主分析流程不啟動它們。

此頁保留既有教學連結，原本的 `WORKFLOW_ACTION` 分支已由三份筆記本取代。
